```python
import yfinance as yf
import pandas as pd

etfs = [
    "ESGU","ESGD","ESGE","DSI","SUSA","USCL","PABU","ESML","ICLN",
    "LCTU","USXF","CRBN","SUSL","DMXF","XVV","XJH","LCTD","PABD","SDG","EMXF"
]

data_full = yf.download(
    etfs,
    start="2017-01-01",
    end="2025-12-31",
    auto_adjust=False
)

adj = data_full["Adj Close"]

output_path = "/Users/nityaarya/Downloads/blackrock-esg-etf-study/blackrock-esg-etf-study/Data/Processed Data/etf_prices.xlsx"

with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
    adj.to_excel(writer, sheet_name="ETF_Prices")
    
    workbook  = writer.book
    worksheet = writer.sheets["ETF_Prices"]

    worksheet.set_column(1, adj.shape[1], 15)  
    worksheet.set_column(0, 0, 18)           

    worksheet.freeze_panes(1, 1)

print(f" Saved ETF prices to {output_path}")
```

```python
import yfinance as yf
import pandas as pd

benchmarks = ["^GSPC", "ACWI", "URTH", "^RUT"]

bench_data = yf.download(
    benchmarks,
    start="2017-01-01",
    end="2025-12-31",
    auto_adjust=False
)["Adj Close"]

output_path = "/Users/nityaarya/Downloads/blackrock-esg-etf-study/blackrock-esg-etf-study/Data/Processed Data/benchmark_prices.xlsx"
with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
    bench_data.to_excel(writer, sheet_name="Benchmarks")

    workbook  = writer.book
    worksheet = writer.sheets["Benchmarks"]

    worksheet.set_column(0, 0, 18)   # Date column
    worksheet.set_column(1, len(benchmarks), 15)

    worksheet.freeze_panes(1, 1)

print(f" Saved benchmark prices (with Russell 2000) to {output_path}")
print(bench_data.head())
```
