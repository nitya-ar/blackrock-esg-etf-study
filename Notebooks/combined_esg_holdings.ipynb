```python
import pandas as pd
import glob
import os

raw_data_path = "/Users/nityaarya/Downloads/blackrock-esg-etf-study/Data/Raw Data/All Holdings 2025"
output_path = "/Users/nityaarya/Downloads/blackrock-esg-etf-study/Data/Processed Data/holdings_2025.csv"

file_paths = glob.glob(os.path.join(raw_data_path, "*_holdings.xlsx"))

column_mapping = {
    'ticker': 'Ticker',
    'name': 'Name',
    'sector': 'Sector',
    'asset class': 'Asset Class',
    'market value': 'Market Value',
    'weight (%)': 'Weight (%)',
    'location': 'Location',
    'exchange': 'Exchange',
    'currency': 'Currency'
}

combined_data = []

for file in file_paths:
    try:
        etf_ticker = os.path.basename(file).split("_")[0]
        print(f"\nProcessing file: {etf_ticker}")

        df = pd.read_excel(file, sheet_name=0, header=10)
        print("File loaded successfully.")

        df.columns = [str(col).strip().lower() for col in df.columns]
        print("Original columns:", df.columns.tolist())

        df = df.rename(columns=column_mapping)
        print("Renamed columns:", df.columns.tolist())

        if 'Name' not in df.columns:
            print(f"Skipping {file} – 'Name' column missing after rename.")
            continue

        if 'Ticker' not in df.columns:
            df['Ticker'] = None

        df = df[df['Name'].notna()]
        df['ETF_Ticker'] = etf_ticker

        expected_cols = ['ETF_Ticker'] + list(column_mapping.values())
        for col in expected_cols:
            if col not in df.columns:
                df[col] = None

        df = df[expected_cols]

        combined_data.append(df)
        print(f"Successfully processed: {etf_ticker} ({len(df)} rows)")

    except Exception as e:
        print(f"Error processing {file}: {e}")

if combined_data:
    final_df = pd.concat(combined_data, ignore_index=True)
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    final_df.to_csv(output_path, index=False)
    print(f"\n holdings_2025 saved to:\n{output_path}")
else:
    print("\n No valid holdings files were processed.")
```
